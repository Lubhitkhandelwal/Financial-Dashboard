```python
import pandas as pd
import numpy as np
from statsmodels.tsa.arima.model import ARIMA
# import statsmodels.api as sm
import os
import plotly.graph_objects as go
from sklearn.metrics import mean_absolute_error, mean_squared_error

# --- Paths ---
DATA_PATH = '../data/processed/cleaned_finance_data.csv'
MODEL_PATH = '../models/forecaster/arima_model.pkl'
MODEL_DIR = '../models/forecaster/'

os.makedirs(MODEL_DIR, exist_ok=True)
```

```python
# --- 1. Load Data ---
df = pd.read_csv(DATA_PATH, parse_dates=['record_date'])
print("Data loaded.")

# --- 2. Prepare Time Series Data ---
print("Preparing time series data...")
df_time = df.set_index('record_date')
monthly_avg_expenses = df_time['monthly_expenses_usd'].resample('MS').mean()
monthly_avg_expenses = monthly_avg_expenses.asfreq('MS', fill_value=monthly_avg_expenses.mean())

print("Monthly Average Expenses Time Series:")
print(monthly_avg_expenses.head(10))
```

```python
# --- 3. Split Data into Train and Test Sets ---
SPLIT_DATE = '2025-01-01'
train_data = monthly_avg_expenses[monthly_avg_expenses.index <= SPLIT_DATE]
test_data = monthly_avg_expenses[monthly_avg_expenses.index > SPLIT_DATE]

print(f"\nTraining data from {train_data.index.min()} to {train_data.index.max()} ({len(train_data)} points)")
print(f"Testing data from {test_data.index.min()} to {test_data.index.max()} ({len(test_data)} points)")

# --- 4. Build & Train Model on TRAINING data only ---
print("\nTraining ARIMA model on TRAIN data...")
model = ARIMA(train_data, order=(1, 1, 1))
model_fit = model.fit()

print(model_fit.summary())
```

```python
# --- 5. Make Validation AND Future Predictions ---
print("\nGenerating validation and future predictions...")
n_steps_test = len(test_data)
n_steps_future = 12 # How many months into the future to predict
n_steps_total = n_steps_test + n_steps_future

# Get the forecast object for the total steps
forecast_object = model_fit.get_forecast(steps=n_steps_total)

# Get the mean predictions
all_predictions = forecast_object.predicted_mean

# Get the 95% confidence interval DataFrame
all_conf_int_df = forecast_object.conf_int(alpha=0.05)
all_lower_bounds = all_conf_int_df.iloc[:, 0]
all_upper_bounds = all_conf_int_df.iloc[:, 1]

# --- Split the forecast into "validation" and "future" ---

# Validation part (for comparing against test_data)
validation_predictions = all_predictions.iloc[:n_steps_test]
validation_predictions.index = test_data.index # Assign correct test dates

validation_lower = all_lower_bounds.iloc[:n_steps_test]
validation_lower.index = test_data.index
validation_upper = all_upper_bounds.iloc[:n_steps_test]
validation_upper.index = test_data.index

# Future part (new predictions)
future_predictions = all_predictions.iloc[n_steps_test:]
future_lower = all_lower_bounds.iloc[n_steps_test:]
future_upper = all_upper_bounds.iloc[n_steps_test:]
```

```python
# --- 6. Calculate Validation Metrics (on test data only) ---
mae = mean_absolute_error(test_data, validation_predictions)
mse = mean_squared_error(test_data, validation_predictions)
rmse = np.sqrt(mse)

print("\n--- Model Validation Metrics (Backtest) ---")
print(f"Mean Absolute Error (MAE): ${mae:,.2f}")
print(f"Root Mean Squared Error (RMSE): ${rmse:,.2f}")
print(f"\nThis means our model's backtest was, on average, off by ${mae:,.2f}.")

# --- 7. NEW: Plot Everything ---
print("Generating full validation and forecast plot...")
fig = go.Figure()

# Plot Validation Confidence Interval (light green)
fig.add_trace(go.Scatter(
    x=validation_upper.index, y=validation_upper, mode='lines',
    line=dict(color='rgba(170, 170, 170, 0)'), showlegend=False
))
fig.add_trace(go.Scatter(
    x=validation_lower.index, y=validation_lower, mode='lines',
    line=dict(color='rgba(170, 170, 170, 0)'), fill='tonexty',
    fillcolor='rgba(0, 100, 80, 0.15)', name='95% Confidence Interval (Validation)'
))

# Plot Future Confidence Interval (light purple)
fig.add_trace(go.Scatter(
    x=future_upper.index, y=future_upper, mode='lines',
    line=dict(color='rgba(170, 170, 170, 0)'), showlegend=False
))
fig.add_trace(go.Scatter(
    x=future_lower.index, y=future_lower, mode='lines',
    line=dict(color='rgba(170, 170, 170, 0)'), fill='tonexty',
    fillcolor='rgba(100, 0, 100, 0.1)', name='95% Confidence Interval (Future)'
))

# Plot training data (blue)
fig.add_trace(go.Scatter(
    x=train_data.index, y=train_data.values, mode='lines',
    name='Historical Training Data', line=dict(color='#0068C9')
))

# Plot *actual* test data (orange)
fig.add_trace(go.Scatter(
    x=test_data.index, y=test_data.values, mode='lines',
    name='Actual Data (Test Set)', line=dict(width=3, color='#FF8200')
))

# Plot *validation* prediction (dotted red)
fig.add_trace(go.Scatter(
    x=validation_predictions.index, y=validation_predictions.values, mode='lines',
    name='Validation Predictions', line=dict(dash='dot', color='#D40000', width=2)
))

# Plot *future* prediction (dotted purple)
fig.add_trace(go.Scatter(
    x=future_predictions.index, y=future_predictions.values, mode='lines',
    name='Future Forecast', line=dict(dash='dot', color='#5800D4', width=2)
))

fig.update_layout(
    title="ARIMA Model Validation & Future Forecast",
    xaxis_title="Date",
    yaxis_title="Average Monthly Expense (USD)",
    legend_title="Data Type"
)
fig.show()
```

```python
# --- 8. FINAL: Re-train Model on ALL Data for the App ---
print("\nValidation complete. Re-training model on FULL dataset for the app...")
final_model = ARIMA(monthly_avg_expenses, order=(1, 1, 1))
final_model_fit = final_model.fit()

# Save the FINAL model
final_model_fit.save(MODEL_PATH)
print(f"\nFinal ARIMA model (trained on all data) saved to {MODEL_PATH}")
```
