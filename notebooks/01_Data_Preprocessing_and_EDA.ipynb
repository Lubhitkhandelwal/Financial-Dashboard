```python
import pandas as pd
import numpy as np
import plotly.express as px
import os

# --- 1. Load Data ---
# Make sure your file is named this, or change the name
RAW_DATA_PATH = '../data/raw/Personal-Finance-ML-Dataset.csv' 
PROCESSED_DATA_PATH = '../data/processed/cleaned_finance_data.csv'
PROCESSED_DIR = '../data/processed/'

# Create the processed directory if it doesn't exist
os.makedirs(PROCESSED_DIR, exist_ok=True)

df = pd.read_csv(RAW_DATA_PATH)
print("Data loaded successfully.")
```

```python
# --- 2. Clean Data ---
print("\nCleaning data...")

# Standardize column names
df.columns = df.columns.str.lower().str.strip().str.replace(' ', '_')

print("Standardized column names:")
print(df.columns.to_list())

# We now use 'record_date' as our date column
if 'record_date' not in df.columns:
    print("---------------------------------------------------------")
    print("ERROR: The 'record_date' column is still not found.")
    print("Please check your CSV file. Available columns are:")
    print(df.columns.to_list())
    print("---------------------------------------------------------")
else:
    # Convert 'record_date' to datetime objects
    df['record_date'] = pd.to_datetime(df['record_date'])

    # Check for missing values
    print(f"\nMissing values:\n{df.isnull().sum()}")
    # This dataset has no missing values, but if it did, you'd use:
    # df.fillna(0, inplace=True) # or other imputation

    print("Data cleaning complete.")
    print(df.head())
```

```python

# --- 3. Explore Data (EDA) ---
print("\nStarting EDA...")

# We will analyze 'monthly_expenses_usd'
# Line chart of average expenses over time
df_time = df.set_index('record_date')
monthly_avg_expenses = df_time['monthly_expenses_usd'].resample('MS').mean().reset_index()

fig1 = px.line(monthly_avg_expenses, 
            x='record_date', 
            y='monthly_expenses_usd', 
            title='Average Monthly Expenses Over Time')
fig1.show()

# Histogram of Credit Scores
fig2 = px.histogram(df, x='credit_score', title='Distribution of Credit Scores')
fig2.show()

# Savings vs. Expenses
fig3 = px.scatter(df.sample(1000), # Sample for performance
                x='monthly_expenses_usd',
                y='savings_usd',
                color='education_level',
                title='Savings vs. Monthly Expenses')
fig3.show()

print("EDA complete. Check the plots.")
```

```python
# --- 4. Save Processed Data ---
df.to_csv(PROCESSED_DATA_PATH, index=False)
print(f"\nProcessed data saved to {PROCESSED_DATA_PATH}")
```
